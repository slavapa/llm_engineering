# Day 1 Exercise — Email Subject Line Generator

**Personal exercise** (fork only, not for upstream PR)

Build a small commercial-style feature: given an email body, suggest a short subject line — like Gmail Smart Compose or Outlook suggestions.

Uses the same Chat Completions pattern from `day1.ipynb`.

```python
import os

from dotenv import load_dotenv
from openai import OpenAI

load_dotenv(override=True)
openai = OpenAI()
```

## Step 1: Define prompts

```python
system_prompt = """
You are an assistant that reads email drafts and suggests a concise subject line.
The subject should be clear, professional, and under 60 characters when possible.
Respond with only the subject line — no quotes, no explanation.
"""

user_prompt_prefix = "Suggest a subject line for this email:\n\n"
```

## Step 2: Sample emails (themed around our exercise sites)

```python
sample_emails = {
    "kabuconnect_event": """
Hi everyone,

This Sunday we have a live Q&A event with Joseph Ohayon and Tony Kosinec.
You can ask anything and connect with fellow students on KabU Connect.
All the details are on kabuconnect.com — hope to see you there!

Best,
The KabU Team
""",
    "kabbalahmedia_newsletter": """
Dear subscriber,

New lectures and articles have been added to Kabbalah Media this week,
including translated materials in English. Visit kabbalahmedia.info/en/
to explore the latest content on the wisdom of Kabbalah.

Warm regards,
Kabbalah Media
""",
    "michaellaitman_article": """
Hello,

Dr. Michael Laitman published a new article on michaellaitman.com about
connection in today's world and practical steps for personal growth.
I thought you might find it relevant — let me know if you'd like the link.

Thanks,
Alex
""",
}
```

## Step 3: Build messages and call OpenAI

```python
def messages_for(email_body: str) -> list[dict]:
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + email_body.strip()},
    ]


def suggest_subject_line(email_body: str, model: str = "gpt-4.1-mini") -> str:
    response = openai.chat.completions.create(
        model=model,
        messages=messages_for(email_body),
    )
    return response.choices[0].message.content.strip()
```

## Step 4: Try it on the sample emails

```python
for name, email in sample_emails.items():
    subject = suggest_subject_line(email)
    print(f"--- {name} ---")
    print(f"Subject: {subject}")
    print()
```
